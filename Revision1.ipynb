```python
import numpy as np
scores = np.array([
    45, 67, 72, 88, 56,
    91, 63, 78, 84, 52,
    69, 73, 95, 61, 87
])
     ## TASK 1

# 1. Finding the number of students
numberOfStudents=np.shape(scores)
print(f'Number of Students: ',numberOfStudents)
print('--------------------------------------------------------------')
# 2. Highest Score
highestScore=np.max(scores)
print(f'Highest Score: ',highestScore)
print('--------------------------------------------------------------')
# 3. Lowest Score
lowestScore=np.min(scores)
print(f'Lowest Score: ',lowestScore)
print('--------------------------------------------------------------')
# 4. Total Score
totalScore=np.sum(scores)
print(f'Total Score: ',totalScore)
print('--------------------------------------------------------------')
# 5. Average Score
averageScore=round(np.average(scores),2)
print(f'Average Score: ',averageScore)
```

```python
                        # TASK 2
                        # Descriptive Statistics
# 6. Calculating median
median=np.median(scores)
print(f'Median: ',median)
print('--------------------------------------------------------------')
# 7. Variance
variance=round(np.var(scores),2)
print(f'Variance: ',variance)
print('--------------------------------------------------------------')
# 8. Standard Deviation
standardDeviation=round(np.std(scores),2)
print(f'Standard Deviation: ',standardDeviation)
print('--------------------------------------------------------------')
# 9. Range
range=np.max(scores)-np.min(scores)
print(f'Range: ',range)
print('--------------------------------------------------------------')
# 10. 25th Percentile
percentile25=np.percentile(scores, 25)
print(f'25th Percentile: ',percentile25)
print('--------------------------------------------------------------')
percentile75=np.percentile(scores,75)
print(f'75TH Percentile: ',percentile75)
```

```python
# Boolean Filtering
# 11. Student who scored above 70
above70=scores[scores>70]
print(f'Student Who Scored above 70: ',above70)
print('---------------------------------------------------')
# 12. Students who scored below 60
below60=scores[scores<60]
print(f'Students who scored below 60: ',below60)
print('---------------------------------------------------')
# 13. Students who scored between 60 and 80
between60and80=scores[(scores>=60) & (scores<=80)]
print(f' Student who scored between 60 and 80: ',between60and80)
```
